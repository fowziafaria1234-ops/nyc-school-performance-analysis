# NYC Public School Test Results Analysis

An MI-focused investigation of performance patterns across 350+ school-year records.

> **Data note:** This portfolio reconstruction uses a reproducible synthetic dataset because the original training files were not available for publication.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
ROOT=Path("..").resolve()
df=pd.read_csv(ROOT/"data/processed/nyc_school_results_clean.csv")
df.head()
```

## Borough ranking

```python
borough=df.groupby("borough").agg(median_composite=("composite_score","median"),median_attendance=("attendance_rate_pct","median"),schools=("school_id","nunique")).sort_values("median_composite",ascending=False)
borough.round(2)
```

## Data-quality evidence

```python
pd.Series({"rows":len(df),"missing_required":df[["math_score","reading_score","writing_score","attendance_rate_pct"]].isna().sum().sum(),"duplicates":df.duplicated(["school_id","year"]).sum(),"outliers":int(df.outlier_flag.sum())})
```

## Relationship between attendance and performance

```python
corr=df[["attendance_rate_pct","composite_score"]].corr().iloc[0,1]
print(f"Correlation: {corr:.3f}")
df.plot.scatter(x="attendance_rate_pct",y="composite_score",alpha=.45,figsize=(10,5),title="Attendance and composite performance")
plt.tight_layout();
```
